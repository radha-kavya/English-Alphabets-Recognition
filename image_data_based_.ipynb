```python
# Converting All images into Numpy array
import numpy as np

image_num_data = []

for i in range(len(image_paths)):
    image = cv2.imread(image_paths[i])

    # Resize the image
    image = cv2.resize(
        src=image,
        dsize=(32, 32),
        interpolation=cv2.INTER_LINEAR
    )

    image_num = np.array(image)

    # Normalize pixel values
    image_flat = image_num.flatten() / 255.0

    image_num_data.append(image_flat)
```

```python
# converting list into numpy array
image_array = np.array(image_num_data)
```

```python
# Verifying the properties of numpy array
print(type(image_array))
print(image_array.shape)
print(image_array.ndim)
```

```python
# Converting numpy array into Data Frame
df = pd.DataFrame(image_array)
```

```python
df.head()
```

```python
# Add labels to the dataset
labels = []
image_num_data = []

for path in image_paths:

    image = cv2.imread(path)
    image = cv2.resize(image, (32,32))

    image_flat = image.flatten()
    image_num_data.append(image_flat)

    # Extract folder name
    label = path.split("/")[-2]
    labels.append(label)
```

```python
import matplotlib.pyplot as plt
import cv2

classes = sorted(set(labels))

plt.figure(figsize=(15,10))

for i, cls in enumerate(classes):
    for path in image_paths:
        if path.split("/")[-2] == cls:
            image = cv2.imread(path)
            image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

            plt.subplot(4,7,i+1)
            plt.imshow(image)
            plt.title(cls)
            plt.axis("off")
            break

plt.suptitle("Sample Images from Each Class", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
# Seperate the features X and y
y = df["Alphabets"]
X = df.drop(columns="Alphabets")

print(X.shape)
print(y.shape)
```

```python
# Split the data
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,shuffle=True,stratify=y,random_state=42)

print(X_train.shape)
print(X_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
X_train.head()
```

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.tree import DecisionTreeClassifier

from sklearn.metrics import accuracy_score

# Dictionary of models
models = {
    "KNN": KNeighborsClassifier(n_neighbors=5),
    "Naive Bayes": GaussianNB(),
    "Decision Tree": DecisionTreeClassifier(random_state=42)

}

# Train and Evaluate
for name, model in models.items():

    model.fit(X_train, y_train)

    y_pred = model.predict(X_test)

    acc = accuracy_score(y_test, y_pred)

    print("="*40)
    print(f"Model : {name}")
    print(f"Accuracy : ",acc)
```

```python
# Hyper Parameter Tuning for KNN
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import RandomizedSearchCV

knn = KNeighborsClassifier()

param_dist = {
    "n_neighbors": [1, 3, 5, 7, 9, 11, 13, 15],
    "weights": ["uniform", "distance"],
    "metric": ["euclidean", "manhattan"]
}

random_search = RandomizedSearchCV(
    estimator=knn,
    param_distributions=param_dist,
    n_iter=5,
    cv=3,
    scoring="accuracy",
    n_jobs=-1,
    random_state=42
)

random_search.fit(X_train, y_train)

print("Best Parameters:", random_search.best_params_)
print("Best CV Accuracy:", random_search.best_score_)
```

```python
best_knn = random_search.best_estimator_

y_pred = best_knn.predict(X_test)

print("Tuned Test Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# Print Classification Report
from sklearn.metrics import classification_report,confusion_matrix

print(classification_report(y_test, y_pred))
```

```python
# Print Confusion Matrix
print( confusion_matrix(y_test, y_pred))
```
